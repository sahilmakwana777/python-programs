```python
# Tuples and Sets in Python

# Tuple
numbers = (10, 20, 30, 40, 50)

print("Tuple:", numbers)
print("First element:", numbers[0])
print("Last element:", numbers[-1])
print("Tuple length:", len(numbers))

# Set
myset = {10, 20, 30, 40}

print("Set:", myset)

# Adding an element
myset.add(50)
print("After adding 50:", myset)

# Removing an element
myset.remove(20)
print("After removing 20:", myset)

# Set operations
set1 = {1, 2, 3}
set2 = {3, 4, 5}

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
