```python
# Program to explain Mutable and Immutable objects in Python

# Immutable object: Integer
a = 10
print("Before changing immutable object:")
print("a =", a)
print("ID of a:", id(a))

a = 20
print("\nAfter changing a:")
print("a =", a)
print("ID of a:", id(a))


# Mutable object: List
my_list = [10, 20, 30]
print("\nBefore changing mutable object:")
print("my_list =", my_list)
print("ID of my_list:", id(my_list))

my_list.append(40)
print("\nAfter changing my_list:")
print("my_list =", my_list)
print("ID of my_list:", id(my_list))
```
