```python
# Create a dictionary
student = {
    "name": "Sahil",
    "age": 22,
    "course": "MCA"
}

print("Original Dictionary:")
print(student)

# Dictionary methods
print("\nKeys:")
print(student.keys())

print("\nValues:")
print(student.values())

print("\nItems:")
print(student.items())

# Adding a new item
student["city"] = "Rajkot"

print("\nAfter adding city:")
print(student)

# Iteration through dictionary
print("\nDictionary elements:")
for key in student:
    print(key, ":", student[key])
```
