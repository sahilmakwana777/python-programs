```python
# Program to demonstrate break, continue and pass

print("Break Statement")

for i in range(1, 6):
    if i == 4:
        break
    print(i)


print("Continue Statement")

for i in range(1, 6):
    if i == 3:
        continue
    print(i)


print("Pass Statement")

for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```
