```python
# User-defined function with arguments

def add(a, b):
    print("Addition =", a + b)

# Positional arguments
add(10, 20)

# Keyword arguments
add(b=30, a=15)


# Function with default argument
def greet(name="Sahil"):
    print("Hello", name)

greet()
greet("Rahul")


# Function with variable number of arguments
def total(*numbers):
    sum = 0
    for n in numbers:
        sum = sum + n
    print("Total =", sum)

total(10, 20, 30)
```
