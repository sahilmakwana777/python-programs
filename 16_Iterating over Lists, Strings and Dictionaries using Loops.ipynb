```python
# Program to iterate over list, string and dictionary

# List
numbers = [10, 20, 30, 40, 50]

print("List elements:")

for number in numbers:
    print(number)


# String
name = "Sahil"

print("String characters:")

for letter in name:
    print(letter)


# Dictionary
student = {
    "name": "Sahil",
    "age": 22,
    "course": "MCA"
}

print("Dictionary elements:")

for key, value in student.items():
    print(key, ":", value)
```
