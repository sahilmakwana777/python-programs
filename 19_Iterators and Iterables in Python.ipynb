```python
# Program to demonstrate iterable and iterator

numbers = [10, 20, 30, 40, 50]

# List is an iterable
print("Iterable:")

for number in numbers:
    print(number)


# Convert iterable into iterator
numbers_iterator = iter(numbers)

print("Iterator:")

print(next(numbers_iterator))
print(next(numbers_iterator))
print(next(numbers_iterator))
print(next(numbers_iterator))
print(next(numbers_iterator))
```
